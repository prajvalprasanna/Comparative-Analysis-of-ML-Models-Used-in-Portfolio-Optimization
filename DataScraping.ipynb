```python
import yfinance as yf
import pandas as pd
import time
```

### Scraping data of companies from different sectors
1. Technology
2. Finance
3. Energy
4. Consumer Goods
5. Healthcare
6. Automobiles
7. Metals and Mining
8. Real Estate


```python
# List of NSE Tech stock symbols to fetch data for
stock_symbols = [
# IT Sector
    'TCS.NS',
    'INFY.NS',
    'WIPRO.NS',
    'HCLTECH.NS',
    'TECHM.NS',
    'LTI.NS',
    'MINDTREE.NS',
    'MPHASIS.NS',
    'COFORGE.NS',
    'PERSISTENT.NS'
]

# Start date and end date -- time period to extract historical data
start_date = '2014-04-01'
end_date = '2024-03-31'

# Dictionary to hold data for each stock
stock_data = {}

# Loop through each stock symbol and fetch data
for symbol in stock_symbols:
    print(f"Fetching data for {symbol}...")
    try:
        # Fetch historical data
        stock = yf.Ticker(symbol)
        data = stock.history(start = start_date, end = end_date)  # Fetch historical data from the last 5 years
        stock_data[symbol] = data
        print(f"Successfully fetched data for {symbol}.")
        
        # Fetch additional fundamental data
        fundamentals = stock.info
        print(f"Fundamental data for {symbol}:")
        print(fundamentals)
        
        # Wait for 2 seconds to avoid overloading the server
        time.sleep(2)
    except Exception as e:
        print(f"Error fetching data for {symbol}: {e}")

# Combine all the stock data into a single DataFrame
combined_data = pd.DataFrame()

# Align the data by date (outer join)
for symbol in stock_symbols:
    if symbol in stock_data:
        # Rename columns to include the stock symbol as a prefix
        stock_data[symbol].columns = [f"{symbol}_{col}" for col in stock_data[symbol].columns]
        if combined_data.empty:
            combined_data = stock_data[symbol]
        else:
            combined_data = combined_data.join(stock_data[symbol], how='outer')

# Display the combined data
print(combined_data.head())

# Save the combined data to a CSV file
combined_data.to_csv('C:/Users/prajv/Projects/MAHE/Data/nse_tech_stocks_yfinance_data.csv')

# For simplicity, let's fetch the fundamentals into a separate DataFrame
fundamental_data = {symbol: yf.Ticker(symbol).info for symbol in stock_symbols}

# Convert to DataFrame and save to CSV
fundamental_df = pd.DataFrame.from_dict(fundamental_data, orient='index')
fundamental_df.to_csv('C:/Users/prajv/Projects/MAHE/Data/nse_tech_stocks_fundamental_data.csv')
```
